```python

import math

def f(x):
    return math.exp(-x**2)

def trapeze_composite(f, a, b, n):
    h = (b - a) / n
    S = f(a) + f(b)

    for i in range(1, n):
        x = a + i * h
        S += 2 * f(x)

    return h * S / 2


a = 0
b = 1

for n in [4, 6]:
    result = trapeze_composite(f, a, b, n)
    print("n =", n, " I =", result)
```

```python

import math
def f(x):
    return math.exp(-x**2)

def simpson_composite(f, a, b, n):
    h = (b - a) / n
    S = f(a)

    for i in range(1, n):
        x = a + i * h

        if i % 2 == 1:
            S += 4 * f(x)
        else:
            S += 2 * f(x)

    S += f(b)
    return h * S / 3
a = 0
b = 1
for n in [4, 6]:
    result = simpson_composite(f, a, b, n)
    print("n =", n, " I =", result)
```
